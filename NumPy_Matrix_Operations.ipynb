```python

import numpy as np

A = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

B = np.array([
    [7, 8, 9],
    [2, 3, 4]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)

```

```python
print("\nMatrix Addition (A + B):")
print(A + B)

```

```python
print("\nMatrix Subtraction (A - B):")
print(A - B)



```

```python
print("\nElement-wise Multiplication (A * B):")
print(A * B)

print("\nMatrix Multiplication (A @ B.T):")
print(A @ B.T)
```

```python
print("\nTranspose of Matrix A:")
print(A.T)

print("\nTranspose of Matrix B:")
print(B.T)
```
